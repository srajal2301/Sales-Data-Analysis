```python
# Calculate key metrics
total_sales = df["sales"].sum()
total_profit = df["profit"].sum()
profit_margin = (total_profit / total_sales) * 100

category_sales = df.groupby("category")["sales"].sum()
category_profit = df.groupby("category")["profit"].sum()
region_profit = df.groupby("region")["profit"].sum()

monthly_sales = df.groupby(
    df["order_date"].dt.to_period("M")
)["sales"].sum()

product_profit = df.groupby("product_name")["profit"].sum()
order_sales = df.groupby("order_id")["sales"].sum()

# Create project summary
summary = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Total Profit",
        "Overall Profit Margin (%)",
        "Total Unique Orders",
        "Total Unique Customers",
        "Total Unique Products",
        "Highest-Selling Category",
        "Most Profitable Category",
        "Most Profitable Region",
        "Best Sales Month",
        "Loss-Making Products",
        "Average Sales per Order"
    ],
    "Value": [
        round(total_sales, 2),
        round(total_profit, 2),
        round(profit_margin, 2),
        df["order_id"].nunique(),
        df["customer_name"].nunique(),
        df["product_name"].nunique(),
        category_sales.idxmax(),
        category_profit.idxmax(),
        region_profit.idxmax(),
        str(monthly_sales.idxmax()),
        int((product_profit < 0).sum()),
        round(order_sales.mean(), 2)
    ]
})

print(summary.to_string(index=False))
```
